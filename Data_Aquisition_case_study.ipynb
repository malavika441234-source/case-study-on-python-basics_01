## Task 8
Merge all four datasets into one DataFrame.

Hint:

● Use the common column (country)

● Choose an appropriate join method.

```python
merged_df = population.merge(df_gdp, on='country', how='outer') \
               .merge(net_users, on='country', how='outer') \
               .merge(df_lit, on='country', how='outer')
```

```python
merged_df
```

## Task 9
Create the following calculated columns:

1. Internet Penetration (%)

2. GDP Per Capita

Hint:

Round the calculated values to two decimal places.

```python
merged_df["internet_penetration (%)"] =(merged_df["internet_users"] / merged_df["population"] * 100)
```

```python
merged_df["GDP_Per_Capita"] = (merged_df["GDP"] * 1000000 / merged_df["population"])
```

##Data Analysis Using SQL

11. Display the Top 5 countries by GDP.

```python
# Display the Top 5 countries by GDP. using sqlite
conn=sqlite3.connect('country_statistics.db')

pd.read_sql("""
SELECT country, GDP
FROM country_statistics
ORDER BY GDP DESC
LIMIT 5
""", conn)
```

12. Find the average literacy rate.


```python
pd.read_sql("""
SELECT AVG(literacy_rate) AS
average_litracy_rate
from country_statistics
""", conn)
```

13. Display all countries having literacy rate above 90%.


```python
pd.read_sql(
    """
SELECT country, literacy_rate
FROM country_statistics
WHERE literacy_rate > 90
ORDER BY literacy_rate DESC
""",
    conn,
)
```

14. Find the Top 5 countries with the highest internet penetration.


```python
pd.read_sql(
    """
SELECT country, `internet_penetration (%)`
FROM country_statistics
ORDER BY `internet_penetration (%)` DESC
LIMIT 5
""",
    conn,
)
```

15. #Calculate the total population.


```python
pd.read_sql(
    """
SELECT SUM(population) AS total_population
FROM country_statistics
""",
    conn,
)
```

16. Find the country with the highest GDP per capita.


```python
pd.read_sql(
    """
SELECT country, GDP_Per_Capita
FROM country_statistics
ORDER BY GDP_Per_Capita DESC
LIMIT 1
""",
    conn,
)
```

17. Export the final merged dataset as a CSV file


```python
df = pd.read_sql("SELECT * FROM country_statistics", conn)

df.to_csv("final_merged_dataset.csv", index=False)

print("CSV file exported successfully!")
```
