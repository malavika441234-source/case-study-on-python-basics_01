```python
# handling outliers in sqft_living
```

```python
# Outlier handling using IQR method

q1 = house_df['sqft_living'].quantile(0.25)
q2 = house_df['sqft_living'].quantile(0.50)
q3 = house_df['sqft_living'].quantile(0.75)

print(f'Q1 = {q1}\nQ2 = {q2}\nQ3 = {q3}')

iqr = q3 - q1

up_lim = q3 + (1.5 * iqr)
low_lim = q1 - (1.5 * iqr)

print(f'IQR = {iqr}\nUp_lim = {up_lim}\nLow_lim = {low_lim}')
```

```python
house_df.loc[house_df['sqft_living'] > up_lim,'sqft_living'] = up_lim

house_df.loc[house_df['sqft_living'] < low_lim,'sqft_living'] = low_lim
```

# splitting and scaling

```python
x= house_df.drop('price',axis=1)
y= house_df['price']
```

```python
x1 =x.copy()
```

```python
from sklearn.preprocessing import StandardScaler
scaler= StandardScaler()
x1 = scaler.fit_transform(x1)
```

```python
# scaled x_train1 and x_test1
from sklearn.model_selection import train_test_split
x_train1, x_test1, y_train, y_test = train_test_split(x1, y, test_size=0.2, random_state=42)
```

```python
# non scaled x_train and x_test
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

# Linear Regression

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score,mean_absolute_error
lr = LinearRegression()
model = lr.fit(x_train1,y_train)
y_pred = model.predict(x_test1)
#using MSR, r2_score matrices for model evaluation
print('mean_squared_error',mean_squared_error(y_test,y_pred))
print('r2_score',r2_score(y_test,y_pred))
print('mean_absolute_error',mean_absolute_error(y_test,y_pred))
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score,mean_absolute_error
lr = LinearRegression()
model = lr.fit(x_train,y_train)
y_pred = model.predict(x_test)
#using MSR, r2_score matrices for model evaluation
print('mean_squared_error',mean_squared_error(y_test,y_pred))
print('r2_score',r2_score(y_test,y_pred))
print('mean_absolute_error',mean_absolute_error(y_test,y_pred))
```
